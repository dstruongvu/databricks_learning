```python
movies_df = spark.read.csv("/Volumes/workspace/default/study_test/movies.csv", header=True, inferSchema=True)
movies_df.show(5)
```

```python
movies_df.printSchema()
```

```python
movies_df.createOrReplaceTempView("movies")
```

```python
%sql
select primaryTitle, rank, averageRating, numVotes, runtimeMinutes, genres, IMDbLink, Title_IMDb_Link 
from movies
```

```python
%sql
select genres, round(avg(averageRating), 2) as avg_rating
from movies
group by genres
order by avg_rating desc
limit 10
```

```python
from pyspark.sql.functions import avg, round

movies_df.groupBy("genres") \
    .agg(round(avg("averageRating"), 2).alias("avg_rating")) \
    .orderBy("avg_rating", ascending=False) \
    .show(10)
```

```python
%sql
select primaryTitle, averageRating
from movies
where averageRating > 7
```
