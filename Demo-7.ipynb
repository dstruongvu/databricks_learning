```python
movies_df = spark.read.csv("/Volumes/workspace/thinknyx/thinknyx_vol/movies.csv",
                           header=True,
                           inferSchema=True)
movies_df.show()
movies_df.printSchema()
```

# Check number of partitions distributions

```python
from pyspark.sql.functions import spark_partition_id

movies_with_pid = movies_df.withColumn("partition_id", spark_partition_id())
display(movies_with_pid)
```

```python
movies_with_pid.select('partition_id').distinct().show()
```

# Repartition

```python
repartitioned_df = movies_df.repartition(5)
```

```python
repartitioned_with_pid = repartitioned_df.withColumn("partition_id", spark_partition_id())
display(repartitioned_with_pid)
```

```python
repartitioned_with_pid.select('partition_id').distinct().show()
```

```python
repartitioned_df.explain(True)
```

# Coalesce

```python
coalesce_df = repartitioned_df.coalesce(2)
```

```python
coalesce_with_pid = coalesce_df.withColumn("partition_id", spark_partition_id())
display(coalesce_with_pid)
```

```python
coalesce_with_pid.select("partition_id").distinct().show()
```
